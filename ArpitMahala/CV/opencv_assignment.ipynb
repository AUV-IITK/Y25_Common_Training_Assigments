## 2. Drawing and Text
**Task:**
1. Create a blank image of size $500 \times 500$ with 3 color channels.
2. Paint a portion of the image red ($[200:300, 300:400]$).
3. Draw a green rectangle from $(0,0)$ to the center of the image.
4. Draw a red circle at the center with a radius of 40.
5. Draw a white line from $(100, 250)$ to $(300, 400)$.



```python
import numpy as np

image2 = np.zeros((500, 500, 3), dtype='uint8')

image2[200:300, 300:400] = (0, 0, 255)
# Blue = 0, Green = 0, Red = 255

cv2.rectangle(image2, (0, 0), (250, 250), (0, 255, 0), thickness=2)

cv2.circle(image2, (250, 250), 40, (0, 0, 255), thickness=2)

cv2.line(image2, (100,250), (300, 400), (255, 255, 255), thickness = 2)

image_rgb = cv2.cvtColor(image2, cv2.COLOR_BGR2RGB)

plt.imshow(image_rgb)
plt.axis('off')
plt.show()
```

```python
image3 = cv2.imread('park.jpg')

grayimage = cv2.cvtColor(image3, cv2.COLOR_BGR2GRAY)

blurred_image = cv2.GaussianBlur(grayimage, (7, 7), 0)

edges = cv2.Canny(blurred_image, 50, 70)

plt.imshow(edges, cmap='gray')
plt.axis('off')
plt.show()
```

## 4. Image Transformations
**Task:**
1. Create a function `translate(img, x, y)` to shift an image.
2. Create a function `rotate(img, angle, rotPoint=None)` to rotate an image.
3. Apply a flip transformation to `park.jpg`.

```python
# Write your code here
def translate(image, x, y):

    m, n = image.shape[:2]
    
    M = np.float32([[1, 0, x], 
                    [0, 1, y]])
    
    return cv2.warpAffine(image, M, (n, m))

def rotate(image, angle, rotPoint = None) :
    m, n = image.shape[:2]

    if rotPoint is None:
        rotPoint = (n // 2, m // 2)

    M = cv2.getRotationMatrix2D(rotPoint, angle, scale=1.0)

    return cv2.warpAffine(image, M, (n, m))

        

image4 = cv2.flip(image3, 1)

image_rgb = cv2.cvtColor(image4, cv2.COLOR_BGR2RGB)

plt.imshow(image_rgb)
plt.axis('off')
plt.show()
```

## 5. Contour Detection
**Task:**
1. Load `cats.jpg`.
2. Convert to grayscale and apply Gaussian blur.
3. Find the edges using Canny.
4. Use `cv.findContours` to find all contours.
5. Print the count and draw them on a blank image.

```python
# Write your code here
image_rgb = cv2.imread('cat.jpg')

image5 = cv2.cvtColor(image_rgb, cv2.COLOR_BGR2GRAY)
image5 = cv2.GaussianBlur(image5, (7, 7), 0)

image5 = cv2.Canny(image5, 50, 80)

contours, hierarchy = cv2.findContours(image5, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

print(f"Total contours found: {len(contours)}")

output_img = image_rgb.copy()
cv2.drawContours(output_img, contours, -1, (0, 255, 0), 3)


plt.imshow(output_img)
plt.axis('off')
plt.show()
```

## 6. Color Spaces
**Task:**
1. Load `park.jpg`.
2. Convert and display in: Gray, HSV,and RGB.

```python
# Write your code here
image = cv2.imread('park.jpg')

grayimage = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
hsvimage = cv2.cvtColor(image, cv2.COLOR_BGR2HSV)
rgbimage = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

plt.subplot(1, 3, 1)
plt.imshow(grayimage, cmap='gray')
plt.title('Grayscale Image')
plt.axis('off')

# --- Plot 2: HSV ---
plt.subplot(1, 3, 2)
plt.imshow(hsvimage)
plt.title('HSV Image')
plt.axis('off')

# --- Plot 3: RGB ---
plt.subplot(1, 3, 3)
plt.imshow(rgbimage)
plt.title('RGB Image')
plt.axis('off')

plt.tight_layout()
plt.show()
```

## 7. Smoothing and Blurring
**Task:**
1. Load `cats.jpg`.
2. Apply Averaging, Gaussian and Median Blurs

```python
# Write your code here
image = cv2.imread('cat.jpg')

gaussianimg = cv2.GaussianBlur(image, (7, 7), 0)
averageimg = cv2.blur(image, (5,5))
medianimg = cv2.medianBlur(image, 5)

gaussianimg = cv2.cvtColor(gaussianimg, cv2.COLOR_BGR2RGB)
averageimg = cv2.cvtColor(averageimg, cv2.COLOR_BGR2RGB)
medianimg = cv2.cvtColor(medianimg, cv2.COLOR_BGR2RGB)


plt.subplot(1, 3, 1)
plt.imshow(averageimg)
plt.title('Averaging')
plt.axis('off')


plt.subplot(1, 3, 2)
plt.imshow(gaussianimg)
plt.title('Gaussian Blur')
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(medianimg)
plt.title('Median Blur')
plt.axis('off')

plt.tight_layout()
plt.show()
```

## 8. Bitwise Operations
**Task:**
1. Create a $400 \times 400$ blank image.
2. Create a rectangle and a circle on separate blank images.
3. Perform Bitwise AND, OR, XOR, and NOT.

```python
# Write your code here
image1 = np.zeros((400, 400), dtype='uint8')
image2 = np.zeros((400, 400), dtype='uint8')

cv2.rectangle(image1, (100,100), (300,300), 255, thickness= -1)
cv2.circle(image2, (200,200), 50, 255, thickness= -1)

plt.subplot(1, 4, 1)
plt.imshow(image1 & image2, cmap='gray')
plt.title('bitwise and')
plt.axis('off')


plt.subplot(1, 4, 2)
plt.imshow(image1 | image2, cmap='gray')
plt.title('bitwise or')
plt.axis('off')


plt.subplot(1, 4, 3)
plt.imshow(image1 ^ image2, cmap='gray')
plt.title('bitwise xor')
plt.axis('off')

plt.subplot(1, 4, 4)
plt.imshow(~image2, cmap='gray')
plt.title('bitwise not of circle')
plt.axis('off')
plt.tight_layout()
plt.show()
```

## 9. Thresholding
**Task:**
1. Load `cats.jpg` and convert to grayscale.
2. Apply Simple Thresholding (Binary and Inverse).

```python
# Write your code here
image = cv2.imread('cat.jpg')
image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

_, bin = cv2.threshold(image, 100, 255, cv2.THRESH_BINARY)
_, inv = cv2.threshold(image, 100, 255, cv2.THRESH_BINARY_INV)

plt.subplot(1, 2, 1)
plt.imshow(bin, cmap='gray')
plt.title('binary thresholding')
plt.axis('off')


plt.subplot(1, 2, 2)
plt.imshow(inv, cmap='gray')
plt.title('inverse binary thresholding')
plt.axis('off')

plt.tight_layout()
plt.show()
```
