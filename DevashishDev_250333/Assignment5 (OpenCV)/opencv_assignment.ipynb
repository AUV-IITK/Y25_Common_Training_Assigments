## 1. Reading Images and Videos
**Task:**
1. Read the image `cats.jpg` and display it.
2. Read the video `dog.mp4` and display it frame by frame. Ensure the video window closes when the 'd' key is pressed.

```python
# Write your code here
import cv2 as cv

#image reading
img = cv.imread('opencv_files/cat.jpg')
cv.imshow('Cat Image', img)

#video reading
capture = cv.VideoCapture('opencv_files/dog.mp4')
while True:
    isTrue, frame = capture.read()

    if not isTrue:
        break

    cv.imshow('Dog Video', frame)

    if cv.waitKey(20) & 0xFF == ord('d'):
        break

capture.release()
cv.destroyAllWindows()
```

## 4. Image Transformations
**Task:**
1. Create a function `translate(img, x, y)` to shift an image.
2. Create a function `rotate(img, angle, rotPoint=None)` to rotate an image.
3. Apply a flip transformation to `park.jpg`.

```python
# Write your code here
import cv2 as cv
import numpy as np

img = cv.imread('opencv_files/park.jpg')

# -----------------------------
# Translation Function
# -----------------------------
def translate(img, x, y):

    # Translation matrix
    transMat = np.float32([
        [1, 0, x],
        [0, 1, y]
    ])

    dimensions = (img.shape[1], img.shape[0])

    return cv.warpAffine(img, transMat, dimensions)


# -----------------------------
# Rotation Function
# -----------------------------
def rotate(img, angle, rotPoint=None):

    (height, width) = img.shape[:2]

    # Rotate around center if no point given
    if rotPoint is None:
        rotPoint = (width // 2, height // 2)

    # Rotation matrix
    rotMat = cv.getRotationMatrix2D(rotPoint, angle, 1.0)

    dimensions = (width, height)

    return cv.warpAffine(img, rotMat, dimensions)


translated = translate(img, 100, 100)

rotated = rotate(img, 45)

flipped = cv.flip(img, 1)

cv.imshow('Original', img)
cv.imshow('Translated', translated)
cv.imshow('Rotated', rotated)
cv.imshow('Flipped', flipped)

cv.waitKey(0)
cv.destroyAllWindows()
```

## 5. Contour Detection
**Task:**
1. Load `cats.jpg`.
2. Convert to grayscale and apply Gaussian blur.
3. Find the edges using Canny.
4. Use `cv.findContours` to find all contours.
5. Print the count and draw them on a blank image.

```python
# Write your code here
import cv2 as cv
import numpy as np

img = cv.imread('opencv_files/cat.jpg')

gray = cv.cvtColor(img, cv.COLOR_BGR2GRAY)

# Apply Gaussian Blur
blur = cv.GaussianBlur(gray, (5, 5), 0)

# Perform Canny Edge Detection
canny = cv.Canny(blur, 125, 175)

# Find contours
contours, hierarchies = cv.findContours(
    canny,
    cv.RETR_LIST,
    cv.CHAIN_APPROX_SIMPLE
)

# Print number of contours
print(f'{len(contours)} contour(s) found.')

# Create blank image
blank = np.zeros(img.shape, dtype='uint8')

# Draw contours on blank image
cv.drawContours(blank, contours, -1, (0, 255, 0), 1)

cv.imshow('Original Image', img)
cv.imshow('Canny Edges', canny)
cv.imshow('Contours', blank)

cv.waitKey(0)
cv.destroyAllWindows()
```

## 8. Bitwise Operations
**Task:**
1. Create a $400 \times 400$ blank image.
2. Create a rectangle and a circle on separate blank images.
3. Perform Bitwise AND, OR, XOR, and NOT.

```python
# Write your code here
import cv2 as cv
import numpy as np

# Create a 400x400 blank image
blank = np.zeros((400, 400), dtype='uint8')

# Create rectangle on separate blank image
rectangle = cv.rectangle(blank.copy(),
                         (30, 30),
                         (370, 370),
                         255,
                         -1)

# Create circle on separate blank image
circle = cv.circle(blank.copy(),
                   (200, 200),
                   200,
                   255,
                   -1)

bitwise_and = cv.bitwise_and(rectangle, circle)

bitwise_or = cv.bitwise_or(rectangle, circle)

bitwise_xor = cv.bitwise_xor(rectangle, circle)

bitwise_not = cv.bitwise_not(circle)

cv.imshow('Rectangle', rectangle)
cv.imshow('Circle', circle)
cv.imshow('AND', bitwise_and)
cv.imshow('OR', bitwise_or)
cv.imshow('XOR', bitwise_xor)
cv.imshow('NOT', bitwise_not)

cv.waitKey(0)
cv.destroyAllWindows()
```

## 9. Thresholding
**Task:**
1. Load `cats.jpg` and convert to grayscale.
2. Apply Simple Thresholding (Binary and Inverse).

```python
# Write your code here
import cv2 as cv

img = cv.imread('opencv_files/cat.jpg')

gray = cv.cvtColor(img, cv.COLOR_BGR2GRAY)

threshold, thresh = cv.threshold(gray, 150, 255, cv.THRESH_BINARY)

threshold, thresh_inv = cv.threshold(gray, 150, 255, cv.THRESH_BINARY_INV)

cv.imshow('Original Image', img)
cv.imshow('Gray Image', gray)
cv.imshow('Binary Threshold', thresh)
cv.imshow('Inverse Binary Threshold', thresh_inv)

cv.waitKey(5000)
cv.destroyAllWindows()
```
