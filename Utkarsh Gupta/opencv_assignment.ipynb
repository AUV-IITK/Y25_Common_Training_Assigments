## 4. Image Transformations
**Task:**
1. Create a function `translate(img, x, y)` to shift an image.
2. Create a function `rotate(img, angle, rotPoint=None)` to rotate an image.
3. Apply a flip transformation to `park.jpg`.

```python
# Write your code here
import cv2
import numpy as np 
def translate(img,x,y):
    transMat=np.float32([[1,0,x],[0,1,y]])
    dimensions=(img.shape[1],img.shape[0])
    return cv2.warpAffine(img, transMat, dimensions)
def rotate(img,angle,rotPoint=None):
    (height,width)=img.shape[:2]
    if rotPoint is None:
       rotPoint=(width// 2,height//2 )    
    rotMat=cv2.getRotationMatrix2D(rotPoint,angle,1.0)
    dimension=(width,height)
    return cv2.warpAffine(img,rotMat,dimension)
img=cv2.imread("park.jpg")
shifted=translate(img,100,100)
rotated=rotate(img,-45)
flipped =cv2.flip(img,1)
cv2.imshow("shifted image",shifted)
cv2.imshow("rotated image",rotated)       
cv2.imshow("flipped image",flipped)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

## 5. Contour Detection
**Task:**
1. Load `cats.jpg`.
2. Convert to grayscale and apply Gaussian blur.
3. Find the edges using Canny.
4. Use `cv.findContours` to find all contours.
5. Print the count and draw them on a blank image.

```python
# Write your code here
import cv2
import numpy as np
img=cv2.imread("cat.jpg")
gray=cv2.cvtColor(img,cv2.COLOR_BGR2GRAY)
blur=cv2.GaussianBlur(gray,(5,5),cv2.BORDER_DEFAULT)
canny=cv2.Canny(blur,125,175)
contours,hierarchies=cv2.findContours(canny,cv2.RETR_LIST,cv2.CHAIN_APPROX_SIMPLE)
print(f"TOtal countours found:{len(contours)}")
blank=np.zeros(img.shape,dtype='uint8')
cv2.drawContours(blank,contours,-1,(0,0,255),thickness=1)
cv2.imshow("all countours",blank)
cv2.waitKey(0)
cv2.destroyAllWindows()

```

## 7. Smoothing and Blurring

```python
# Write your code here
import cv2
img = cv2.imread('cat.jpg')
average = cv2.blur(img, (7, 7))
gaussian = cv2.GaussianBlur(img, (7, 7), 0)
median = cv2.medianBlur(img, 7)
cv2.imshow("Original", img)
cv2.imshow("Average Blur", average)
cv2.imshow("Gaussian Blur", gaussian)
cv2.imshow("Median Blur", median)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

## 8. Bitwise Operations
**Task:**
1. Create a $400 \times 400$ blank image.
2. Create a rectangle and a circle on separate blank images.
3. Perform Bitwise AND, OR, XOR, and NOT.

```python
# Write your import cv2
import numpy as np
blank = np.zeros((400, 400), dtype='uint8')
rectangle = blank.copy()
cv2.rectangle(rectangle, (30, 30), (370, 370), 255, thickness=-1)
circle = blank.copy()
cv2.circle(circle, (200, 200), 200, 255, thickness=-1)
bitwise_and = cv2.bitwise_and(rectangle, circle)
bitwise_or = cv2.bitwise_or(rectangle, circle)
bitwise_xor = cv2.bitwise_xor(rectangle, circle)
bitwise_not = cv2.bitwise_not(circle)
cv2.imshow("Rectangle", rectangle)
cv2.imshow("Circle", circle)
cv2.imshow("Bitwise AND", bitwise_and)
cv2.imshow("Bitwise OR", bitwise_or)
cv2.imshow("Bitwise XOR", bitwise_xor)
cv2.imshow("Bitwise NOT (Circle)", bitwise_not)
cv2.waitKey(0)
cv2.destroyAllWindows()

```

## 9. Thresholding

```python
# Write your code here
import cv2
import numpy as np
img = cv2.imread('cat.jpg')
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
threshold_value, thresh = cv2.threshold(gray, 125, 255, cv2.THRESH_BINARY)
threshold_value_inv, thresh_inv = cv2.threshold(gray, 125, 255, cv2.THRESH_BINARY_INV)
cv2.imshow("Original Grayscale", gray)
cv2.imshow("Simple Thresholding (Binary)", thresh)
cv2.imshow("Inverse Thresholding", thresh_inv)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
