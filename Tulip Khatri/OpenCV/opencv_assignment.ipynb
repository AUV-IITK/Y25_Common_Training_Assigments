## 4. Image Transformations
**Task:**
1. Create a function `translate(img, x, y)` to shift an image.
2. Create a function `rotate(img, angle, rotPoint=None)` to rotate an image.
3. Apply a flip transformation to `park.jpg`.

```python
import cv2
import numpy as np
from matplotlib import pyplot as plt
image2= cv2.imread("park.jpg")

def translate(image2,x,y):
  transMat= np.float32([[1,0,x],[0,1,y]])
  dimensions=(image2.shape[1], image2.shape[0])
  return cv2.warpAffine(image2, transMat, dimensions)

def rotate(image2, angle, rotPoint=None):
  (height, width) = image2.shape[:2]
  if rotPoint is None:
    rotPoint = (width // 2, height // 2)

  rotMat = cv2.getRotationMatrix2D(rotPoint, angle, 1.0)
  dimensions = (width, height)

  return cv2.warpAffine(image2, rotMat, dimensions)

rotated_image = rotate(image2, 180)
rotated_image_rgb = cv2.cvtColor(rotated_image, cv2.COLOR_BGR2RGB)
plt.imshow(rotated_image_rgb)
plt.axis("off")
plt.show()
```

## 5. Contour Detection
**Task:**
1. Load `cats.jpg`.
2. Convert to grayscale and apply Gaussian blur.
3. Find the edges using Canny.
4. Use `cv.findContours` to find all contours.
5. Print the count and draw them on a blank image.

```python

import cv2
import matplotlib.pyplot as plt

image = cv2.imread("cat.jpg")
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
gray_cat = cv2.cvtColor(image, cv2.COLOR_RGB2GRAY)
blur_cat = cv2.GaussianBlur(gray_cat, (7, 7), 0)
edges = cv2.Canny(blur_cat, 100, 180)
contours, _ = cv2.findContours(edges, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

# print the count and draw them on a blank image
print("Number of contours:", len(contours))
blank = np.zeros_like(image)
cv2.drawContours(blank, contours, -1, (255, 34, 0), 1)
plt.imshow(blank)
```

## 7. Smoothing and Blurring
**Task:**
1. Load `cats.jpg`.
2. Apply Averaging, Gaussian and Median Blurs

```python

import cv2
import matplotlib.pyplot as plt

image = cv2.imread("cat.jpg")
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

#averaging
average = cv2.blur(image, (3, 3))
plt.imshow(average)
plt.axis("off")
plt.show()

#gaussian
gaussian = cv2.GaussianBlur(image, (3, 3), 0)
plt.imshow(gaussian)
plt.axis("off")
plt.show()

#median
median = cv2.medianBlur(image, 3)
plt.imshow(median)
plt.axis("off")
plt.show()

```

## 8. Bitwise Operations
**Task:**
1. Create a $400 \times 400$ blank image.
2. Create a rectangle and a circle on separate blank images.
3. Perform Bitwise AND, OR, XOR, and NOT.

```python

import cv2
import numpy as np
import matplotlib.pyplot as plt
img = np.zeros((400, 400, 3),dtype=np.uint8)
cv2.rectangle(img, (0, 0), (250, 250), (0, 255, 0), 3)
cv2.circle(img, (250, 250), 40, (0, 0, 255), -1)

#perform bitwise and,or,xor and not
bitwise_and = cv2.bitwise_and(img, img)
bitwise_or = cv2.bitwise_or(img, img)
bitwise_xor = cv2.bitwise_xor(img, img)
bitwise_not = cv2.bitwise_not(img)

plt.imshow(bitwise_and)
plt.axis("off")
plt.show()

plt.imshow(bitwise_or)
plt.axis("off")
plt.show()

plt.imshow(bitwise_xor)
plt.axis("off")
plt.show()

plt.imshow(bitwise_not)
plt.axis("off")
plt.show()


```

## 9. Thresholding
**Task:**
1. Load `cats.jpg` and convert to grayscale.
2. Apply Simple Thresholding (Binary and Inverse).

```python
# Write your code here
import cv2
import matplotlib.pyplot as plt

image = cv2.imread("cat.jpg")
image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
#apply thresholding(binary and inverse )
_, thresh = cv2.threshold(image, 125, 255, cv2.THRESH_BINARY)
_, thresh_inv = cv2.threshold(image, 125, 255, cv2.THRESH_BINARY_INV)

plt.imshow(thresh, cmap="gray")
plt.axis("off")
plt.show()

plt.imshow(thresh_inv, cmap="gray")
plt.axis("off")
plt.show()

```
