## 1. Reading Images and Videos
**Task:**
1. Read the image `cats.jpg` and display it.
2. Read the video `dog.mp4` and display it frame by frame. Ensure the video window closes when the 'd' key is pressed.

```python
import cv2

cat = cv2.imread('opencv_files/cat.jpg')
cv2.imshow('cat', cat)
cv2.waitKey(0)
cv2.destroyAllWindows()
dog = cv2.VideoCapture('opencv_files/dog.mp4')
while True:
    isTrue, frame = dog.read()    
    if not isTrue:
        break     
    cv2.imshow('dog', frame)
    if cv2.waitKey(20) & 0xFF == ord('d'):
        break
dog.release()
cv2.destroyAllWindows()
```

## 3. Basic Image Functions
**Task:**

3. Apply Gaussian Blur with a $(7,7)$ kernel.
4. Perform Canny Edge detection on the blurred image.


```python
import cv2

park = cv2.imread('opencv_files/park.jpg')
gray = cv2.cvtColor(park, cv2.COLOR_BGR2GRAY)
gaus = cv2.GaussianBlur(gray, (7, 7), 0)
canny = cv2.Canny(gaus, 50, 150)
cv2.imshow('Canny Edges', canny)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

## 4. Image Transformations
**Task:**
1. Create a function `translate(img, x, y)` to shift an image.
2. Create a function `rotate(img, angle, rotPoint=None)` to rotate an image.
3. Apply a flip transformation to `park.jpg`.

```python
import cv2
import numpy as np

def translate(img, x, y):
    translation_matrix = np.float32([[1, 0, x], [0, 1, y]])
    dimensions = (img.shape[1], img.shape[0])
    return cv2.warpAffine(img, translation_matrix, dimensions)

def rotate(img, angle, rotPoint=None):
    (height, width) = img.shape[:2]
    if rotPoint is None:
        rotPoint = (width // 2, height // 2)
    rotation_matrix = cv2.getRotationMatrix2D(rotPoint, angle, 1.0)
    dimensions = (width, height)
    return cv2.warpAffine(img, rotation_matrix, dimensions)

park = cv2.imread('opencv_files/park.jpg')
flip = cv2.flip(park, 1)
transflip = translate(flip, 100, 100)
cv2.imshow('Translated Flip', transflip)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

## 5. Contour Detection
**Task:**
1. Load `cats.jpg`.
2. Convert to grayscale and apply Gaussian blur.
3. Find the edges using Canny.
4. Use `cv.findContours` to find all contours.
5. Print the count and draw them on a blank image.

```python
import cv2
import numpy as np

canny = cv2.Canny(cv2.GaussianBlur(cv2.cvtColor(cv2.imread('opencv_files/park.jpg'), cv2.COLOR_BGR2GRAY), (7, 7), 0), 50, 150)
contours, hierarchies = cv2.findContours(canny, cv2.RETR_LIST, cv2.CHAIN_APPROX_SIMPLE)
print(f'n_contours = {len(contours)}')
blank = np.zeros((canny.shape[0], canny.shape[1], 3), dtype='uint8')
cv2.drawContours(blank, contours, -1, (255, 255, 0), thickness=1)
cv2.imshow('Contours', blank)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

## 7. Smoothing and Blurring
